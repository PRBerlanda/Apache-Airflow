```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession\
    .builder\
    .appName("twitter_silver")\
    .getOrCreate()
    
```

```python
df_tweet = spark.read.json("../datalake/Silver/twitter_datascience/tweet/")
df_tweet.printSchema()
```

```python
from pyspark.sql import functions as f
```

```python
tweet_conversas = df_tweet.alias("tweet")\
    .groupBy(f.to_date("created_at").alias("created_date"))\
    .agg(
        f.count("id").alias("n_tweets"),
        f.sum("like_count").alias("n_likes"),
        f.sum("quote_count").alias("n_quotes"),
        f.sum("reply_count").alias("n_replies"),
        f.sum("retweet_count").alias("n_retweets")
    ).withColumn("weekday", f.date_format("created_date", "E"))
```

```python
tweet_conversas.show()
```
