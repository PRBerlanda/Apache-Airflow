```python
import yfinance
```

```python
hist = yfinance\
    .Ticker("AAPL")\
    .history(
        interval= "1h", # 1-hour candles
        start= "2026-08-01", # start date
        end= "2026-09-18", # end date (exclusive)
        prepost=True # include pre- and post-market trading
    )

print(hist.head())
```

```python
def extrai_dados(ticker, start_date, end_date):
    caminho = f"../sample_data/notebook/{ticker}.csv"
    yfinance\
    .Ticker(ticker)\
    .history(
        interval= "1h", # 1-hour candles
        start= start_date, # start date
        end= end_date, # end date (exclusive)
        prepost=True # include pre- and post-market trading
    ).to_csv(caminho)
```

```python
extrai_dados("AAPL", "2026-08-01", "2026-09-18")
```
